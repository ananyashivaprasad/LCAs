```python
student_list=["Evara", "Sitara", "Deirdre", "Darlene"]
student_tuple=("Evara", "Sitara", "Deirdre", "Darlene")
student_dictionary={101:"Evara", 102:"Sitara", 103:"Deirdre", 104:"Darlene"}
print(student_list)
print("\nstudent_tuple")
print(student_tuple)
print("\nstudent_dict")
print(student_dictionary)
```

```python
#Adding an element into dictionary
student_dictionary[105]="Faith"
print("\nAfter Adding:")
print(student_dictionary)
```

```python
#Updating the dictionary
student_dictionary[104]="Akanksha"
print("\nAfter updating:")
print(student_dictionary)
```

```python
#Deleting an element from the dictionary
del student_dictionary[104]
print("\nAfter deleting:")
print(student_dictionary)
```
