```python
import re
def check_string(s):
    pattern = "^[A-Za-z0-9]+$" #^=start, $ = end, +=one or more
    return bool(re.match(pattern, s))
#Example usage
s1="Hello123"
s2="Hello_123"
print(check_string(s1))
print(check_string(s2))
```

```python
def check_string (s):
    return s.isalnum()
#Example usage
s1="Hello123"
s2="Hello_123"
print(check_string(s1))
print(check_string(s2))
```
