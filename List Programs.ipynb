```python
number_list=[21, 22, 24, 26]
print ("the first element is",number_list[0])
print("the second elemeent",number_list[1])
print("the third element is",number_list[2])
print("the fourth element is",number_list[3])
```

```python
#To append numbers
number_list.append(27)
print("\nNumber List")
print(number_list)
```

```python
#To reverse list
number_list.reverse()
print("\nNumber list")
print(number_list)
```

```python
#To print the occurences of the element of the specified list 
numbers_list=[10,10,20,20,30.40]
element=int(input("Enter the number"))
count=numbers_list.count(element)
print("Number of occurences are:", count)
```

```python
#To append the items of list 1 to list 2 in the front 
list1=[10, 20, 40]
list2=[50,60,70]
list1=list2+list1
print("list1:",list1)
print("list2:",list2)
```

```python
#to insert a new elements in the existing list
numbers=[10,20,20,40,50]
numbers.insert(0,90)
print("Numbers")
print(numbers)
```

```python
#to remove an element from specified index
list=[10,20,230,40,50]
index=int(input("Enter index to remove:"))
list.pop(index)
print("After removal of index", list)
```
