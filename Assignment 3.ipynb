```python
#to check if the triangle is a right angled triangle or not (user defined)
a=int(input("Enter base:"))
b=int(input("Enter height:"))
c=int(input("Enter other side"))
def right_angled(a,b,c):
 if a*a+b*b==c*c:
   print("It is a right angled triangle")
 else:
   print("It is not a right angled triangle")
right_angled(a,b,c)
```

```python
#to check if the triangle is right angled or not (pre defined values)
def right_angled(a,b,c):
    if a*a+b*b==c*c:
        print("It is a right angled triangle")
    else:
        print("It is not a right angled triangle")
right_angled(3,4,5)
```
