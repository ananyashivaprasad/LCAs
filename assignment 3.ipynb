```python
#to check if the triangle is a right angle triangle or not (user based)

a=int(input("Enter base:"))
b=int(input("Enter height:"))
c=int(input("Enter other side"))
def right_angle(a,b,c):
 if a*a+b*b==c*c:
  print("Right angled triangle")
 else:
  print("Not a right angled triangle")
right_angle(a,b,c)
```
